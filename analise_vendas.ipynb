```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("vendas.csv")

print(df.head())
print("\n")
print(df.info())

df["faturamento"] = df["quantidade"] * df["preco_unitario"]

faturamento_total = df["faturamento"].sum()
print(f"\n Faturamento total: R$ {faturamento_total:,.2f}")

top_produtos = df.groupby("produto")["quantidade"].sum().sort_values(ascending=False).head(5)
print(f"\n {top_produtos}")

faturamento_categoria = df.groupby("categoria")["faturamento"].sum().sort_values(ascending=False)
print(f"\n {faturamento_categoria}")
print("\n")

faturamento_categoria.plot(kind="bar")
plt.title("Faturamento por Categoria")
plt.ylabel("R$")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
